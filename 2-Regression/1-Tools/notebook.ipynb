```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets,linear_model, model_selection

X, y = datasets.load_diabetes(return_X_y=True)
print(X.shape)
print(X[0])
```

```python
# selecting a portion of dataset, 2nd column is selected

X = X[:, 3]
X= X.reshape(-1, 1)
# print(X)
```

```python
# spliting the dataset into training and testing sets

X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33)
```

```python
# training model, loading linear regression model

model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

```python
# creating prediction using test data

y_pred = model.predict(X_test)


# plotting
plt.scatter(X_test, y_test,  color='black')
plt.plot(X_test, y_pred, color='blue', linewidth=3)
plt.xlabel('Scaled Blood Pressure')
plt.ylabel('Disease Progression')
plt.title('A Graph Plot Showing Diabetes Progression Against Blood Pressure')
plt.show()
```
